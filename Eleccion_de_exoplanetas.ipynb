# Buscamos un TOI
Repetimos el algoritmo anterior

### Creamos el dataframe

````python
tabla = NasaExoplanetArchive.query_criteria(table="toi", select="*")
toi = tabla.to_pandas()
toi.to_csv("toi.csv", index=False)
````

### Mostramos el dataframe

````python
toi = pd.read_csv("toi.csv")
toi.head()
````

### Tomamos los datos que necesitamos

````python
toi_id = toi['tid'].tolist()
toi_num = toi['toi'].tolist()
toi_pnum = toi['pl_pnum'].tolist()
toi_disp = toi['tfopwg_disp'].tolist()
````

## Contamos los cuerpos de cada estrella
Cada fila de la tabla es un TOI, es decir, un posible planeta. Nos quedamos con los planetas confirmados (**CP**, **KP**) y los candidatos (**PC**, **APC**), y descartamos los falsos positivos (**FP**) y las falsas alarmas (**FA**).

- **CP**: planeta confirmado
- **KP**: planeta ya conocido
- **PC**: Planet Candidate
- **APC**: Ambiguous Planet Candidate

No se usa la columna **pl_pnum**, porque su valor cambia entre los TOI de una misma estrella. Por eso los cuerpos se cuentan con un **Counter**.

````python
# numero de planetas deseados en el sistema
numerodeplanetas = 3

# lista te planetas viables
tois = []

for i in range(len(toi_id)):
    if toi_disp[i] in ['CP', 'KP'] or toi_disp[i] in ['PC', 'APC']: # PC: Planet Candidate y APC: Ambiguous Planet Candidate
        tois.append(toi_id[i])

# se guardan todos los toi para determinar los falsos positivos
todos_los_tid = tois
````

analisis de falsos positivos

````python
from collections import Counter

conteo = Counter(todos_los_tid)
````

El **Counter** de la librería **collections** cuenta cuántas veces aparece cada elemento de una lista. Como cada estrella aparece una vez por cada cuerpo, el conteo es el número de cuerpos del sistema.

ejemplo:

```python
from collections import Counter

Counter(['a', 'b', 'a'])   # Counter({'a': 2, 'b': 1})
```

### Sistemas con 2 o 3 cuerpos
Nos quedamos con las estrellas que tienen 2 o 3 cuerpos (confirmados o candidatos).

````python
toi_viables = []

for estrella, veces in conteo.items():
    if 2 <= veces <= 3:
        toi_viables.append(estrella)
````

## Propiedades de las estrellas
Para saber si una estrella sirve para medir velocidades radiales necesitamos su brillo en TESS (**st_tmag**), su temperatura (**st_teff**), su radio (**st_rad**) y su declinación (**dec**).

````python
toi_tmag = toi['st_tmag'].tolist()
toi_teff = toi['st_teff'].tolist()
toi_rad = toi['st_rad'].tolist()
toi_dec = toi['dec'].tolist()
````

### Datos de las estrellas viables
Los datos de la estrella se repiten en cada TOI, así que se guardan una sola vez por estrella.

````python
v_tid = []
v_tmag = []
v_teff = []
v_rad = []
v_dec = []

for i in range(len(toi_id)):
    if toi_id[i] in toi_viables and toi_id[i] not in v_tid:
        v_tid.append(toi_id[i])
        v_tmag.append(toi_tmag[i])
        v_teff.append(toi_teff[i])
        v_rad.append(toi_rad[i])
        v_dec.append(toi_dec[i])
len(v_tid)
````

### Rango de los datos
Vemos el mínimo y el máximo de cada propiedad para elegir los cortes.

````python
print(min(v_tmag), max(v_tmag))
print(min(v_teff), max(v_teff))
print(min(v_dec), max(v_dec))
````

## Cortes por brillo y temperatura
Se buscan enanas M (**Teff < 4000 K**) con brillo menor a 12 en la banda de TESS, para poder medir velocidades radiales con buena precisión. Los cortes se pueden cambiar. Las estrellas que cumplen los dos cortes son las **finalistas**.

````python
# cortes (se pueden cambiar)
tmag_max = 12
teff_max = 4000

n_brillo = 0
n_fria = 0
n_ambas = 0
finalistas = []

for i in range(len(v_tid)):
    brillante = v_tmag[i] < tmag_max
    fria = v_teff[i] < teff_max

    if brillante:
        n_brillo += 1
    if fria:
        n_fria += 1
    if brillante and fria:
        n_ambas += 1
        finalistas.append(v_tid[i])

print(n_brillo, n_fria, n_ambas)
````

### Zona del cielo
Según la declinación se decide qué observatorios pueden ver la estrella:

- **solo sur** (dec < -30°): HARPS, ESPRESSO y NIRPS (Chile)
- **solo norte** (dec > 30°): MAROON-X y SPIRou
- **ambos**: entre -30° y 30°

Los límites son aproximados.

````python
for estrella in finalistas:
    j = v_tid.index(estrella)   # posición de la estrella en las listas v_
    if v_dec[j] < -30:
        zona = 'solo sur'
    elif v_dec[j] > 30:
        zona = 'solo norte'
    else:
        zona = 'ambos'
    print(estrella, conteo[estrella], v_tmag[j], v_teff[j], v_dec[j], zona)
````

# Espectros disponibles
## Probamos la consulta al archivo de ESO
Se usa el servicio TAP del archivo de ESO con **pyvo**. La consulta está escrita en ADQL y cuenta cuántas observaciones hay en la tabla **dbo.raw** dentro de una pequeña caja alrededor de la estrella. Probamos con Próxima Centauri, que tiene cientos de espectros.

Se usa una caja en **ra** y **dec** en lugar de **CIRCLE**, porque algunas filas de la tabla tienen coordenadas inválidas y la consulta falla.

````python
import pyvo
import math

tap = pyvo.dal.TAPService("https://archive.eso.org/tap_obs")

ra = 217.4289
dec = -62.6795
radio = 0.02   # grados

# la caja en RA se ensancha según la declinación
dra = radio / math.cos(math.radians(dec))

consulta = f"""
SELECT instrument, COUNT(*) AS n
FROM dbo.raw
WHERE dp_cat = 'SCIENCE'
  AND instrument IN ('HARPS', 'ESPRESSO', 'NIRPS')
  AND dec BETWEEN {dec - radio} AND {dec + radio}
  AND ra BETWEEN {ra - dra} AND {ra + dra}
GROUP BY instrument
"""

resultado = tap.search(consulta).to_table()
print(resultado)
````

### Espectros de cada finalista
Repetimos la consulta para las estrellas del sur y de "ambos" (declinación hasta 30°) y guardamos cuántos espectros tienen en HARPS, ESPRESSO y NIRPS. Las del norte quedan para después.

````python
toi_ra = toi['ra'].tolist()
toi_num = toi['toi'].tolist()
toi_per = toi['pl_orbper'].tolist()
toi_rade = toi['pl_rade'].tolist()
radio = 0.02


q_tid = []
q_harps = []
q_esp = []
q_nirps = []

for estrella in finalistas:
    j = v_tid.index(estrella)
    dec = v_dec[j]

    if dec <= 30:   # las del norte quedan para después
        i = toi_id.index(estrella)
        ra = toi_ra[i]
        dra = radio / math.cos(math.radians(dec))

        consulta = f"""
        SELECT instrument, COUNT(*) AS n
        FROM dbo.raw
        WHERE dp_cat = 'SCIENCE'
          AND instrument IN ('HARPS', 'ESPRESSO', 'NIRPS')
          AND dec BETWEEN {dec - radio} AND {dec + radio}
          AND ra BETWEEN {ra - dra} AND {ra + dra}
        GROUP BY instrument
        """
        resultado = tap.search(consulta).to_table()

        n_harps = 0
        n_esp = 0
        n_nirps = 0
        for fila in resultado:
            if fila['instrument'] == 'HARPS':
                n_harps = int(fila['n'])
            elif fila['instrument'] == 'ESPRESSO':
                n_esp = int(fila['n'])
            elif fila['instrument'] == 'NIRPS':
                n_nirps = int(fila['n'])

        q_tid.append(estrella)
        q_harps.append(n_harps)
        q_esp.append(n_esp)
        q_nirps.append(n_nirps)

        print(estrella, dec, n_harps, n_esp, n_nirps)
````

### Estrellas con datos en los tres instrumentos
Se muestran los TOI de cada una. Se agrega la estrella 415969908, que corresponde a TOI-233 (el sistema del informe de referencia), para comprobar que coincide.

Resultado: todas las estrellas con datos en los tres instrumentos son planetas ya confirmados (**CP**).

````python
candidatas = []
for k in range(len(q_tid)):
    if q_harps[k] > 0 and q_esp[k] > 0 and q_nirps[k] > 0:
        candidatas.append(q_tid[k])

print(len(candidatas), candidatas)

# estado, periodo y radio de cada TOI (+ 415969908 para comprobar TOI-233)
for estrella in candidatas + [415969908]:
    print('TIC', estrella)
    for i in range(len(toi_id)):
        if toi_id[i] == estrella:
            print('   ', toi_num[i], toi_disp[i], toi_per[i], toi_rade[i])
````

### Estrellas con datos en algunos instrumentos
Se separan las estrellas que tienen espectros en uno o dos instrumentos, sin contar TOI-233.

````python
parciales = []
for k in range(len(q_tid)):
    total = q_harps[k] + q_esp[k] + q_nirps[k]
    if total > 0 and q_tid[k] not in candidatas and q_tid[k] != 415969908:
        parciales.append(q_tid[k])

print(parciales)
````

Vemos el estado de sus TOI. Todas son confirmadas (**CP**) o ya conocidas (**KP**).

````python
for estrella in parciales:
    print('TIC', estrella)
    for i in range(len(toi_id)):
        if toi_id[i] == estrella:
            print('   ', toi_num[i], toi_disp[i], toi_per[i], toi_rade[i])
````

## Buscamos candidatos sin confirmar
Revisamos el resto de las finalistas (las que no tienen datos en ESO y las del norte) y mostramos solo las que tienen al menos un candidato **PC** o **APC**.

````python
for estrella in finalistas:
    if estrella not in candidatas and estrella not in parciales and estrella != 415969908:
        hay_pc = False
        for i in range(len(toi_id)):
            if toi_id[i] == estrella and toi_disp[i] in ['PC', 'APC']:
                hay_pc = True
        if hay_pc:
            print('TIC', estrella)
            for i in range(len(toi_id)):
                if toi_id[i] == estrella:
                    print('   ', toi_num[i], toi_disp[i], toi_per[i], toi_rade[i])
````

# Búsqueda desde ESO
Hasta aquí se partió de los TOI y se preguntó por espectros. Ahora invertimos la búsqueda: se le piden a ESO todas las estrellas con muchos espectros y se cruzan con los TOI candidatos por coordenadas.

## Revisamos la tabla de ESO
Se piden solo 10 filas (**TOP 10**) para ver cómo vienen las columnas. La columna **object** viene vacía, así que no sirve para agrupar por nombre.

````python
consulta = """
SELECT TOP 10 object, instrument, ra, dec
FROM dbo.raw
WHERE dp_cat = 'SCIENCE'
  AND instrument = 'HARPS'
"""
print(tap.search(consulta).to_table())
````

## Espectros por zona del cielo
Como no se puede agrupar por nombre, se agrupan las observaciones por coordenadas redondeadas a 0.01° (unos 36 segundos de arco) y se cuenta cuántas hay de cada instrumento en cada cuadrito.

ADQL solo acepta columnas en el **GROUP BY**, no expresiones como **ROUND(ra, 2)**, por eso el redondeo se hace en una consulta interna.

````python
consulta = """
SELECT instrument, ra2, dec2, COUNT(*) AS n
FROM (
    SELECT instrument, ROUND(ra, 2) AS ra2, ROUND(dec, 2) AS dec2
    FROM dbo.raw
    WHERE dp_cat = 'SCIENCE'
      AND instrument IN ('HARPS', 'ESPRESSO', 'NIRPS')
      AND dec BETWEEN -90 AND 90
      AND ra BETWEEN 0 AND 360
) AS t
GROUP BY instrument, ra2, dec2
HAVING COUNT(*) >= 3
"""

resultado = tap.search(consulta, maxrec=200000).to_table()

e_inst = resultado['instrument'].tolist()
e_ra = resultado['ra2'].tolist()
e_dec = resultado['dec2'].tolist()
e_n = resultado['n'].tolist()

print(len(e_inst))
print(e_inst[:5], e_ra[:5], e_dec[:5], e_n[:5])
````

## Cruce con los TOI
### Estrellas TOI de interés
Se toman las estrellas con 2 o más cuerpos y al menos un candidato (**PC** o **APC**).

### Cruce
Para cada estrella se suman los espectros de los cuadritos que caen a menos de 0.02° y se muestran las que tienen al menos 10.

````python
# 1. estrellas TOI con varios cuerpos y al menos un candidato
min_cuerpos = 2

objetivos = []
o_ra = []
o_dec = []

for estrella, veces in conteo.items():
    if veces >= min_cuerpos:
        hay_pc = False
        for i in range(len(toi_id)):
            if toi_id[i] == estrella and toi_disp[i] in ['PC', 'APC']:
                hay_pc = True
        if hay_pc:
            i = toi_id.index(estrella)
            objetivos.append(estrella)
            o_ra.append(toi_ra[i])
            o_dec.append(toi_dec[i])

print(len(objetivos))

# 2. cruce con los cuadritos de ESO
radio = 0.02     # grados
n_min = 10       # mínimo de espectros en total

for k in range(len(objetivos)):
    n_harps = 0
    n_esp = 0
    n_nirps = 0
    cos_dec = math.cos(math.radians(o_dec[k]))

    for m in range(len(e_inst)):
        if abs(e_dec[m] - o_dec[k]) <= radio and abs(e_ra[m] - o_ra[k]) * cos_dec <= radio:
            if e_inst[m] == 'HARPS':
                n_harps += e_n[m]
            elif e_inst[m] == 'ESPRESSO':
                n_esp += e_n[m]
            elif e_inst[m] == 'NIRPS':
                n_nirps += e_n[m]

    if n_harps + n_esp + n_nirps >= n_min:
        print(objetivos[k], n_harps, n_esp, n_nirps)
````

### Guardamos los conteos
Repetimos el cruce, pero guardando los conteos de cada estrella en listas (**c_tid**, **c_harps**, **c_esp** y **c_nirps**) y sin contar TOI-233.

````python
radio = 0.02
n_min = 10

c_tid = []
c_harps = []
c_esp = []
c_nirps = []

for k in range(len(objetivos)):
    n_harps = 0
    n_esp = 0
    n_nirps = 0
    cos_dec = math.cos(math.radians(o_dec[k]))

    for m in range(len(e_inst)):
        if abs(e_dec[m] - o_dec[k]) <= radio and abs(e_ra[m] - o_ra[k]) * cos_dec <= radio:
            if e_inst[m] == 'HARPS':
                n_harps += e_n[m]
            elif e_inst[m] == 'ESPRESSO':
                n_esp += e_n[m]
            elif e_inst[m] == 'NIRPS':
                n_nirps += e_n[m]

    if n_harps + n_esp + n_nirps >= n_min and objetivos[k] != 415969908:
        c_tid.append(objetivos[k])
        c_harps.append(n_harps)
        c_esp.append(n_esp)
        c_nirps.append(n_nirps)

print(len(c_tid))

for k in range(len(c_tid)):
    estrella = c_tid[k]
    conf = []   # CP y KP
    cand = []   # PC y APC

    for i in range(len(toi_id)):
        if toi_id[i] == estrella:
            if toi_disp[i] in ['CP', 'KP']:
                conf.append(toi_num[i])
            elif toi_disp[i] in ['PC', 'APC']:
                cand.append(toi_num[i])

    i = toi_id.index(estrella)
    print(f"TOI-{int(toi_num[i])} | TIC {estrella} | HARPS {c_harps[k]}, ESPRESSO {c_esp[k]}, NIRPS {c_nirps[k]} | Tmag {toi_tmag[i]:.1f}, Teff {toi_teff[i]:.0f} | CP/KP: {conf} | PC/APC: {cand}")
````

### Separamos los grupos
Las estrellas con algún planeta confirmado (**CP**, **KP**) van a **mixtos_eso** y las demás a **pc_puros_eso**, que son los sistemas con solo candidatos. Después se imprimen con los TOI de cada categoría.

````python
pc_puros_eso = []   # solo candidatos
mixtos_eso = []     # confirmados + candidatos

for estrella in c_tid:
    hay_conf = False
    for i in range(len(toi_id)):
        if toi_id[i] == estrella and toi_disp[i] in ['CP', 'KP']:
            hay_conf = True

    if hay_conf:
        mixtos_eso.append(estrella)
    else:
        pc_puros_eso.append(estrella)

for grupo in [pc_puros_eso, mixtos_eso]:
    for estrella in grupo:
        conf = []   # CP y KP
        cand = []   # PC y APC

        for i in range(len(toi_id)):
            if toi_id[i] == estrella:
                if toi_disp[i] in ['CP', 'KP']:
                    conf.append(toi_num[i])
                elif toi_disp[i] in ['PC', 'APC']:
                    cand.append(toi_num[i])

        i = toi_id.index(estrella)
        print(f"TOI-{int(toi_num[i])} | CP/KP: {conf} | PC/APC: {cand}")
    print()
````

### Excluimos algunos TOI
Se excluyen los TOI de la lista **no**. Se pasan a TIC y se saltan al separar los grupos.

No se modifica **c_tid**, porque va alineada con **c_harps**, **c_esp** y **c_nirps**.

````python
no = ["TOI-214", "TOI-713", "TOI-125", "TOI-201", "TOI-512"]

# pasamos los nombres a TIC
tic_no = []
for nombre in no:
    numero = int(nombre.replace("TOI-", ""))
    for i in range(len(toi_id)):
        if int(toi_num[i]) == numero:
            tic_no.append(toi_id[i])
            break

# separamos los grupos, saltando las estrellas excluidas
pc_puros_eso = []   # solo candidatos
mixtos_eso = []     # confirmados + candidatos

for estrella in c_tid:
    if estrella not in tic_no:
        hay_conf = False
        for i in range(len(toi_id)):
            if toi_id[i] == estrella and toi_disp[i] in ['CP', 'KP']:
                hay_conf = True

        if hay_conf:
            mixtos_eso.append(estrella)
        else:
            pc_puros_eso.append(estrella)

# imprimimos
for grupo in [pc_puros_eso, mixtos_eso]:
    for estrella in grupo:
        conf = []   # CP y KP
        cand = []   # PC y APC

        for i in range(len(toi_id)):
            if toi_id[i] == estrella:
                if toi_disp[i] in ['CP', 'KP']:
                    conf.append(toi_num[i])
                elif toi_disp[i] in ['PC', 'APC']:
                    cand.append(toi_num[i])

        i = toi_id.index(estrella)
        print(f"TOI-{int(toi_num[i])} | CP/KP: {conf} | PC/APC: {cand}")
    print()
````
